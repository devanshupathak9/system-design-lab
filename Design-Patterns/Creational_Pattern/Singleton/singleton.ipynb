# Singleton Design Pattern

**Singleton design pattern** ensures that class has only one object globally.

- *Ensures that class has only one object globally.*
- *Only one instance of the class should exist.*
- *If creating a new object, it will return the older one*.

### Implementation
1. Create a private constructor
2. Create a static instance 
3. Return that same instance everytime 

#### Python Object Creation Lifecycle

When you create an object:

```python
obj = MyClass(*args, **kwargs)
```

Python internally performs the following steps:

```text
MyClass(*args)
      │
      ▼
type(MyClass).__call__()
      │
      ├──► MyClass.__new__(cls, *args, **kwargs)
      │        │
      │        └── Creates and returns a new object
      │
      ├──► MyClass.__init__(self, *args, **kwargs)
      │        │
      │        └── Initializes the object
      │
      ▼
Returns the fully initialized object
```

### Responsibilities

#### __new__(cls, ...)
- Class method (receives `cls`)
- Creates and returns a new instance
- Called **before** `__init__`
- Used for immutable types, Singleton, metaclasses, etc.

#### __init__(self, ...)
- Instance method (receives `self`)
- Initializes the already-created object
- Cannot create or replace the object
- Must return `None`

#### __call__(self, ...)
- Makes an **object callable**
- Invoked when you do:

```python
obj()
```

Example:

```python
class Adder:
    def __call__(self, x, y):
        return x + y

adder = Adder()
adder(2, 3)   # Calls adder.__call__(2, 3)
```

> **Note:** `Class()` invokes the metaclass's `type.__call__()`, which orchestrates the calls to `__new__()` and `__init__()`. `obj()` invokes the object's own `__call__()` method.

1. Ensure that same instance should be accross multiple threads
2. Use lock to prevent creating multiple objects.
3. Use double locking for the thread safety.

````python
class Singleton:
    _instance = None
    def __init__(self):
        pass

    def __new__(cls):
        if cls._instance == None:
            cls._instance = super().__new__(cls)
        return cls._instance
    
instance = Singleton()
````

````python
from threading import Lock, Thread
class Singleton:
    _instance = None
    _lock = Lock()
    def __int__(self):
        pass

    def __new__(cls):
        if cls._instance == None:
            with cls._lock:    
                if cls._instance == None:
                    cls._instance = super().__new__(cls)
        return cls._instance
    
def worker():
    obj = Singleton()
    print(obj)

instance1 = Thread(target=worker)
instance2 = Thread(target=worker)
instance1.start()
instance2.start()
instance1.join()
instance2.join()
````
